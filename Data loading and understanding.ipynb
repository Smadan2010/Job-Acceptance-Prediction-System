```python
# ============================================================
# STEP 2 — DATA LOADING & UNDERSTANDING
# ============================================================

# Import pandas — the main library for working with tables/data
import pandas as pd

# Import numpy — used for math operations
import numpy as np

# ── LOAD THE DATA ──────────────────────────────────────────

# Read the CSV file and store it in a variable called 'df'
# 'df' stands for DataFrame — think of it like an Excel table in Python
df = pd.read_csv("HR_Job_Placement_Dataset.csv")  # ← change path if needed


# ── 1. SHAPE ───────────────────────────────────────────────

# .shape tells us (number of rows, number of columns)
print("Dataset Shape:", df.shape)
# 👉 Output: (51500, 26) → 51,500 candidates, 26 columns


# ── 2. DATA TYPES ──────────────────────────────────────────

# .dtypes shows what TYPE of data is in each column:
#   int64   = whole numbers (like age: 27)
#   float64 = decimal numbers (like score: 78.5)
#   str/obj = text (like "Male", "Placed")
print("\nData Types:")
print(df.dtypes)


# ── 3. FIRST 5 ROWS ────────────────────────────────────────

# .head() shows the first 5 rows — a quick preview of your data
print("\nFirst 5 Rows:")
print(df.head())


# ── 4. LAST 5 ROWS ─────────────────────────────────────────

# .tail() shows the last 5 rows — useful to check if data ends cleanly
print("\nLast 5 Rows:")
print(df.tail())


# ── 5. BASIC STATISTICS ────────────────────────────────────

# .describe() gives count, mean, min, max, etc. for number columns
# This helps us understand the RANGE of values in each column
print("\nBasic Statistics:")
print(df.describe().round(2))


# ── 6. MISSING VALUES ──────────────────────────────────────

# .isnull() marks every blank/missing cell as True
# .sum() counts how many True values (= how many missing) per column
# We also calculate the % to understand how serious it is
print("\nMissing Values:")
missing_count = df.isnull().sum()                        # count of missing
missing_pct   = (missing_count / len(df) * 100).round(2) # percentage missing

# Combine into one neat table
missing_table = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing %'    : missing_pct
})

# Only show columns that actually have missing values
print(missing_table[missing_table['Missing Count'] > 0])


# ── 7. DUPLICATE ROWS ──────────────────────────────────────

# .duplicated() marks rows that are exact copies of another row
# .sum() counts how many such duplicates exist
print("\nDuplicate Rows:", df.duplicated().sum())


# ── 8. CATEGORICAL COLUMNS — UNIQUE VALUES ─────────────────

# select_dtypes picks only text/string columns
# We check each one for unique values — this reveals NOISY DATA
# (e.g., "Yes", "yes", " yes " are all the same but Python sees 3 different things!)

print("\nCategorical Columns — Unique Values:")
cat_cols = df.select_dtypes(include='object').columns  # get text columns

for col in cat_cols:                           # loop through each text column
    print(f"\n[{col}]")                        # print column name
    print(df[col].value_counts())              # print count of each unique value


# ── 9. TARGET VARIABLE DISTRIBUTION ───────────────────────

# How many candidates were Placed vs Not Placed?
print("\nTarget Variable — Status:")
print(df['status'].value_counts())

# Calculate what % of candidates got placed
placed_pct = (df['status'] == 'Placed').mean() * 100
print(f"\nJob Acceptance Rate: {placed_pct:.2f}%")
```
